```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df = pd.read_csv("../data/citations-google-scholar.tsv", sep="\t")
year_cols = ["2019", "2020", "2021", "2022", "2023", "2024", "2025"]
df["label"] = df["title"] + " (" + df["DOI"] + ")"

plot_df = df.set_index("label")[year_cols].T  
plot_df.index = plot_df.index.astype(int)   
```

```python
ax = plot_df.plot(kind="bar", stacked=True)

ax.set_xlabel("Year")
ax.set_ylabel("Number of citations")

ax.set_xticklabels(ax.get_xticklabels(), rotation=0) 

ymax = plot_df.sum(axis=1).max()
ax.set_ylim(0, ymax + 20)
ax.legend(title="")

plt.tight_layout()
plt.savefig("../figs/citations_stacked_by_year.png", dpi=300, bbox_inches="tight")
plt.show()
```
